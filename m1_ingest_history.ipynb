```sql
%python
# M1 - Bronze Ingestion - History
RAW = "/Volumes/workspace/rp_landing/raw"
OPS = "/Volumes/workspace/rp_landing/ops"

from pyspark.sql import functions as F

(spark.readStream.format("cloudFiles")
  .option("cloudFiles.format", "csv")
  .option("header", "true")
  .option("cloudFiles.schemaLocation", f"{OPS}/schemas/history")
  .load(f"{RAW}/landing/history")
  .select("*", F.col("_metadata.file_path").alias("_source_file"),
          F.current_timestamp().alias("_ingest_ts"))
  .writeStream
  .option("checkpointLocation", f"{OPS}/chk/bronze_history")
  .trigger(availableNow=True)
  .toTable("workspace.rp_bronze.history")
)
```

```sql
%python
# M1 - Bronze Ingestion - History (actual folder: sales_history)
RAW = "/Volumes/workspace/rp_landing/raw"
OPS = "/Volumes/workspace/rp_landing/ops"

from pyspark.sql import functions as F

(spark.readStream.format("cloudFiles")
  .option("cloudFiles.format", "csv")
  .option("header", "true")
  .option("cloudFiles.schemaLocation", f"{OPS}/schemas/sales_history")
  .load(f"{RAW}/landing/sales_history")
  .select("*", F.col("_metadata.file_path").alias("_source_file"),
          F.current_timestamp().alias("_ingest_ts"))
  .writeStream
  .option("checkpointLocation", f"{OPS}/chk/bronze_sales_history")
  .trigger(availableNow=True)
  .toTable("workspace.rp_bronze.sales_history")
)
```
