```sql
%python
# M1 - Bronze Ingestion - Products
# Source: /Volumes/workspace/rp_landing/raw/landing/products
# Target: rp_bronze.products

RAW = "/Volumes/workspace/rp_landing/raw"
OPS = "/Volumes/workspace/rp_landing/ops"

from pyspark.sql import functions as F

(spark.readStream.format("cloudFiles")
  .option("cloudFiles.format", "csv")
  .option("header", "true")
  .option("cloudFiles.schemaLocation", f"{OPS}/schemas/products")
  .load(f"{RAW}/landing/products")
  .select("*", F.col("_metadata.file_path").alias("_source_file"),
          F.current_timestamp().alias("_ingest_ts"))
  .writeStream
  .option("checkpointLocation", f"{OPS}/chk/bronze_products")
  .trigger(availableNow=True)
  .toTable("workspace.rp_bronze.products")
)
```

```sql
SELECT count(*) as total_rows, count(DISTINCT product_id) as unique_products 
FROM workspace.rp_bronze.products
```

```sql
SELECT count(*) as total_rows, count(DISTINCT sku) as unique_products 
FROM workspace.rp_bronze.products
```
